```python
weather = input("Weather: ").lower()
temp = int(input("Temperature: "))
if weather == "sunny" and temp >= 25:
    result = "Go to the park."
elif weather == "rainy":
    result = "Choose an indoor activity."
else:
    result = "Take a short walk."
print("AI recommendation:", result)
```
