```python
student=[
    {"name":"bishal","hour":2,"marks":95},
    {"name":"ram","hour":4,"marks":99},
    {"name":"hari","hour":5,"marks":49},
    {"name":"sita","hour":3,"marks":75}
]
for i in student:
    print(i["name"],i["hour"],i["marks"])
```

```python
for i in student:
   if i["marks"]>=60:
    print (i["name"],"pass")
else:
    print(i["name"],"fail")
```

```python
marks=[55,30,60,70,90]
passed=0
fail=0
for mark in marks:
    if mark>=60:
     passed +=1
    else:
     fail+=1
print("passed",passed)
print("failed",fail)
        
        
```
