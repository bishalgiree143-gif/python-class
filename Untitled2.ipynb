```python
from collections import deque
graph = {
"School": ["Station", "Cafe"],
"Station": ["School", "Library"],
"Cafe": ["School", "Park"],
"Library": ["Station", "Hospital"],
"Park": ["Cafe", "Hospital"],
"Hospital": []}

def bfs(start, goal):
    queue = deque([[start]])
    while queue:
        path = queue.popleft()
        node = path[-1]
        print("Expanded:", node)
        if node == goal:
            return path
        for n in graph[node]:
            if n not in path:
                queue.append(path + [n])

print(bfs("School", "Hospital"))
```
