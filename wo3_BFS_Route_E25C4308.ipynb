```python
from collections import deque

graph = {
    "School": ["Library", "Park"],
    "Library": ["School", "Mall"],
    "Park": ["School", "Mall", "Station"],
    "Mall": ["Library", "Park", "Hospital"],
    "Station": ["Park", "Hospital"],
    "Hospital": ["Mall", "Station"]
}
def bfs(graph, start, goal):
    queue = deque([(start, [start])])
    visited = {start}
    removed = []

    while queue:
        node, path = queue.popleft()
        removed.append(node)

        if node == goal:
            return path, removed

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return None, removed

route, expanded = bfs(graph, "School", "Hospital")

print("Route:", " -> ".join(route))
print("Nodes removed from queue:", expanded)

graph["School"].append("Hospital")
graph["Hospital"].append("School")

new_route, new_expanded = bfs(graph, "School", "Hospital")

print("\nNew route:", " -> ".join(new_route))
print("New nodes removed from queue:", new_expanded)
```

#In this code i show old route to new route to go school to hospital.

1.Why did BFS return this route first? BFS explores nodes level by level using a first-in, first-out queue.
#It finds the route with the fewest edges before longer routes in an unweighted graph.
2.How did adding or removing a connection affect the route?
#Adding a direct connection between School and Hospital created a shorter route. Removing a connection could force BFS to find an alternative route or explore more nodes.
3.When is BFS useful or inefficient?
#BFS is useful for finding the shortest path in an unweighted graph. It can be inefficient for large graphs because it may visit many nodes and use considerable memory.
