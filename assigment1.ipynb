```python
mood = input("How is your mood? ").lower()
energy = input("Energy level (high/low): ").lower()

if mood == "happy" and energy == "high":
    result = "Go out with your friends."
elif mood == "sad":
    result = "Listen to your favorite music."
elif energy == "low":
    result = "Take a rest."
else:
    result = "Read a book or learn something new."

print("AI recommendation:", result)
```
