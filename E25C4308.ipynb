```python
import random
rooms = {r: random.choice(["Clean", "Dirty"])
for r in ["A", "B"]}
location, actions = "A", 0
while "Dirty" in rooms.values():
    if rooms[location] == "Dirty":
        rooms[location] = "Clean"
        print(location, "-> CLEAN")
    else:
        location = "B" if location == "A" else "A"
        print("MOVE ->", location)
    actions += 1
print(rooms, "Actions:", actions)
```

```python
import random

random.seed(7)

# Create ten rooms with random conditions
rooms = {
    str(i): random.choice(["Clean", "Dirty"])
    for i in range(1, 11)
}

location = 1
actions = 0

print("Initial rooms:", rooms)

while "Dirty" in rooms.values():

    # Sensor: check the current room
    current_room = str(location)

    if rooms[current_room] == "Dirty":
        # Actuator: clean the room
        rooms[current_room] = "Clean"
        actions += 1
        print("CLEAN -> Room", location)

    else:
        # Find the nearest dirty room
        dirty_rooms = [
            int(room) for room, condition in rooms.items()
            if condition == "Dirty"
        ]

        if dirty_rooms:
            next_room = min(
                dirty_rooms,
                key=lambda room: abs(room - location)
            )

            location = next_room
            actions += 1
            print("MOVE -> Room", location)

print("\nFinal rooms:", rooms)
print("Total actions:", actions)
print("Total cost:", actions, "points")
print("All rooms clean:", all(
    condition == "Clean" for condition in rooms.values()
))
```
